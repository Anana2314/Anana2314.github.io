# Women Accessing Financial Accounts

Over time, women have increasingly gained access to financial accounts. This project will explore whether specific Latin American countries have followed this global trend.

```python
#First dataset, findex from World Bank
import pandas as pd

url_findex = "/Users/anamontanez/Documents/Anana2314.github.io/PROJECT 1/GlobalFindexDatabase2025.csv"
df = pd.read_csv(url_findex)
df.head()
```

```python
#  This dataset have % female population per country (World Bank) 
# because many of the columns at the Findex are percentages but we can't know 
# the number of women to make further comparisions.

url_female = "/Users/anamontanez/Documents/Anana2314.github.io/PROJECT2/Population, female (% of total population).csv"
female_pct = pd.read_csv(url_female)

female_pct.rename(columns={
    "Country Name": "countrynewwb",
    "Year": "year",
    "Value": "pct_female"
}, inplace=True)

```

```python
female_pct.head(5)
```

As a first step it's necessary to merge both datasets because the findex does not have female population information, it only has adult population per country but we need to see how many women have an account and beyond that, how many of them use their bank account and/or have savings.

Also, the WB dataset only has % of women population per country but not the number of women per country.

```python
merged = pd.merge(
    df,
    female_pct[["countrynewwb", "year", "pct_female"]],
    on=["countrynewwb", "year"],
    how="left"
)
```

```python
merged[["countrynewwb", "year", "pop_adult", "pct_female"]].head()
```

```python
#To create the column of n female adult population per country

merged["adult_pop_female"] = merged["pop_adult"] * merged["pct_female"] / 100

merged[["countrynewwb", "year", "pop_adult", "pct_female", "adult_pop_female"]].head()
```

```python
#We will first calculate how many women have an account 
#The variable to use according to their dictionary is 'account_t_d'
#Checking we have all the columns
merged.columns
```

```python
#Filtering women 
women = merged[(merged["group"] == "gender") & (merged["group2"] == "women")].copy()
men    = merged[(merged["group"] == "gender") & (merged["group2"] == "men")].copy()
```

```python
women.head()
women.columns
```

```python
#Now we create women with account

women["women_with_account"] = women["adult_pop_female"] * women["account_t_d"]
```

```python
women[["countrynewwb","year","adult_pop_female","account_t_d","women_with_account"]].head()
```

```python
#Filter only since 2014
women = women[women["year"] >= 2014]
```

```python
#We choose three countries 
target_countries = ["Argentina", "Mexico", "Colombia"]

women = women[women["countrynewwb"].isin(target_countries)]
```

```python
women[["countrynewwb","year","adult_pop_female","account_t_d","women_with_account"]].head()
```

```python
countries = ["Argentina", "Mexico", "Colombia"]
women_sub = women[women["countrynewwb"].isin(countries)].copy()
```

```python
import plotly.express as px

fig = px.line(
    women_sub,
    x="year",
    y="women_with_account",
    color="countrynewwb",
    title="Women with Bank Accounts Over Time",
    markers=True
)

fig.update_layout(
    title={
        "text": "Women with Bank Accounts Over Time",
        "x": 0.5,      
        "xanchor": "center"
    },
    xaxis_title="Year",
    yaxis_title="Women with Account"
)

fig.show()
```

# Conclusion

Mexico is the country with the most sustained growth in the number of women with a bank account. After a slight drop in 2017, it shows a consistent increase from 2017 to 2024, with the most pronounced rise occurring between 2021 and 2024.

Argentina also shows a growing trend starting in 2017, although the pace of growth is more moderate than in Mexico. Even thought, it maintains a continuous increase over the years.

Colombia shows growth between 2014 and 2021, but unlike the other two countries, it experiences a decrease between 2021 and 2024, breaking the previous upward trend.

# Who Gets Arrested in New York? A Data-Driven Look at Race and Crime

Over the past decade, data and research have consistently shown significant racial disparities in New York City’s arrest rates. Black and Hispanic New Yorkers are disproportionately represented among those arrested, relative to their share of the population. Below we summarize key statistical findings to check whether this is still the situation in 2025.

The NYPD has consistently denied engaging in racial profiling, attributing arrest disparities to crime patterns and other factors. Officials state that “the NYPD enforces the law fairly and equally” across all communities. Nonetheless, the statistics and information below evidence unequal impact.

[Data Source Here](https://data.cityofnewyork.us/Public-Safety/NYPD-Arrest-Data-Year-to-Date-/uip8-fykc/about_data)

```python
import pandas as pd

url = "/Users/anamontanez/Documents/Anana2314.github.io/Project_3/NYPD_Arrest_Data_(Year_to_Date)_20251205.csv"

df = pd.read_csv(url)

print(df.head())
```

```python
# Arrest by race in percentages %
conteo_razas = df["PERP_RACE"].value_counts()

porcentaje_razas = (conteo_razas / conteo_razas.sum()) * 100

print(porcentaje_razas.round(2))
```

*From a first basic exploration, we can see that 48% of people arrested are black and 35% are Hispanic.*

```python
print(df.columns)
```

```python
df["ARREST_DATE"] = pd.to_datetime(df["ARREST_DATE"])

df["ARREST_YEAR"] = df["ARREST_DATE"].dt.year
tabla_year = (
    df.groupby(["ARREST_YEAR", "PERP_RACE"])
      .size()
      .groupby(level=0)
      .apply(lambda x: (x / x.sum()) * 100)
      .unstack()
      .round(2)
)

print("% of Arrest by Race:")
print(tabla_year)
```

```python

df["ARREST_MONTH"] = df["ARREST_DATE"].dt.month
tabla_month = (
    df.groupby(["ARREST_MONTH", "PERP_RACE"])
      .size()
      .groupby(level=0)
      .apply(lambda x: (x / x.sum()) * 100)
      .unstack()
      .round(2)
)

print("🎯 % de arrestos por raza por mes (agregado):")
print(tabla_month)
```

## Critical Background and Analysis

1- In 2019, Black individuals made up 48% of NYPD arrests.
Whites accounted for just 11% of arrestees, while Hispanics comprised roughly 34% of arrests.

2- These proportions have remained strikingly consistent in recent years. For example, between 2014 and 2019 (after “stop-and-frisk” was curtailed), 48% of all arrests were of Black people and ~34% were of Hispanics, with whites around 12%.

3- By early 2020 the pattern was similar: NYPD data showed 49% of arrestees were Black, 32% Hispanic, and 11% white

4- In 2025, we can see black population make up by far the highest number of people arrested in the city of New York.

```python
import numpy as np

def convertir_age_group(x):
    if pd.isna(x) or x == "(null)":
        return np.nan
    if x == "<18":
        return 17
    if x == "65+":
        return 70
    # rangos como 25-44
    if "-" in x:
        low, high = x.split("-")
        return (int(low) + int(high)) / 2
    return np.nan

df["AGE_NUM"] = df["AGE_GROUP"].apply(convertir_age_group)
```

Before exploring the offenses by race, we can mentioned that a 2023 analysis by the Brennan Center for Justice found that “low-level” enforcement disparities widened over time. 

In 2016, Black New Yorkers were about 5 times more likely than whites to face minor offense charges; by 2022 they were 6 times more likely. The gap was largest in Manhattan where a Black person was 15 times more likely than a white person to be prosecuted for a minor offense in recent years.

```python
df["OFNS_DESC"].value_counts().head(20)
```

```python
top5_by_race = (
    df.groupby(["PERP_RACE", "OFNS_DESC"])
      .size()
      .reset_index(name="count")
      .sort_values(["PERP_RACE", "count"], ascending=[True, False])
      .groupby("PERP_RACE")
      .head(5)
)
top5_by_race
```

```python
import matplotlib.pyplot as plt

# Obtener lista de razas
races = top5_by_race["PERP_RACE"].unique()

plt.figure(figsize=(16, 12))

for i, race in enumerate(races, 1):
    subset = top5_by_race[top5_by_race["PERP_RACE"] == race]
    
    plt.subplot(3, 3, i) 
    plt.barh(subset["OFNS_DESC"], subset["count"])
    plt.title(race)
    plt.xlabel("Cantidad de arrestos")
    plt.gca().invert_yaxis() 

plt.tight_layout()
plt.show()

plt.tight_layout()   
save_plot("arrests_types_by_race.png") 
plt.show()
```

The distribution of offenses reveals both overlap and meaningful variation across racial groups. Across nearly all groups, Assault 3 & Related Offenses consistently emerges as the most frequent charge, highlighting its prominence in NYPD arrest patterns. Other common offenses—such as Petty Larceny, Dangerous Drugs, and Felony Assault—appear across groups but with different magnitudes, suggesting that while many types of incidents are broadly experienced, their intensity varies by demographic profile. For example, Black and Black Hispanic individuals show especially high counts in assault- and drug-related categories, whereas White and Asian/Pacific Islander individuals show relatively higher shares of property-related offenses like Petty Larceny. 

These differences do not explain causation, but they do illustrate how arrests cluster differently across communities. This type of breakdown underscores the importance of analyzing arrest data disaggregated by race to better understand enforcement patterns, inform policy discussions, and identify areas where further inquiry or targeted intervention may be needed.

# Conclusion

The analysis of NYPD arrest data reveals consistent and meaningful patterns across demographic groups, offense types, and temporal trends. Racial disparities are evident, with certain groups—particularly Black and Hispanic individuals—representing a disproportionate share of arrests across most offense categories. 

When examining the nature of arrests, crimes such as assault, petit larceny, and drug-related offenses emerge as the most common across nearly all racial groups, though the relative distribution varies by demographic profile. Age patterns also reflect concentration in mid-adulthood ranges, while gender differences show that men constitute the overwhelming majority of arrests across all races. Monthly trends remain relatively stable, suggesting that fluctuations in enforcement or criminal activity occur gradually rather than seasonally.

 Taken together, these findings highlight not only broad structural patterns in arrest outcomes but also the importance of considering demographic context when interpreting policing data. This type of disaggregated analysis is essential for informing public policy, evaluating equity in the criminal justice system, and identifying areas where further investigation or intervention may be warranted.
